```python
# Answer - 2
def currency_converter(amount,exchange_rate):
    result = amount * exchange_rate
    return f"{amount} USD is equivalent to {result}PKR"
    
```

```python
exchange_rate = 280
amount = int(input("Enter you USD amount to cenvert into pkr"))
currency_converter(amount,exchange_rate)
```

```python
# Answer - 5
def calculate_rectangle(length, width):
    def area():
        return length * width
    def perimeter():
        return 2 * (length + width)
    print(f"The area is {area()}")
    print(f"The perimeter is {perimeter()}")
    
```

```python
length = int(input("Enter the length of rectangular"))
width = int(input("Enter the width of rectangular"))
calculate_rectangle(length,width)
```

```python
# Answer - 7
def Atm_machine():
    balance = 1000
    try:
        withdraw = int(input("Enter withdraw amount"))
        if withdraw > balance:
            print("Insufficient balance.")
        elif withdraw <= 0:
            print("Amount must be more than zero.")
        else:
            deduction = balance - withdraw
            print(f"Your current balance is {deduction} after deduction {withdraw}")
    except ValueError:
        print("Please enter a valid number.")
    print("Thank you for using the ATM.")
```

```python
Atm_machine()
```

```python
Atm_machine() 
```

```python
Atm_machine() 
```

```python
# Answer - 8
def Booking_tickets():
    seats = 5
    try:
        ticket = int(input("How many tickets do you want to book?"))
        if ticket > seats:
            print("Not enough seats available.")
        elif ticket <= 0:
            print("You must book at least 1 ticket.")
        else:
            print(f"Your {ticket} tickets booked successfully")
            booked = seats -  ticket 
            print(f"Only {booked} seats reamining after your booking {ticket} seats")
    except ValueError:
        print("Please enter a valid number.")
    print("Thank you for booking with us.")
```

```python
Booking_tickets()
```
